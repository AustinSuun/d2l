```python
import matplotlib.pyplot as plt
import numpy as np 
import r2l 
from tqdm import tqdm
import random
import time
```

```python
np.random.seed(0)
random.seed(0)
n_planning_list = [0, 2, 20]
for n_planning in n_planning_list:
    print('Q-learning 步数为：%d' % n_planning)
    time.sleep(0.5)
    return_list = r2l.DynaQ_CliffWalking(n_planning)
    episode_list = list(range(len(return_list)))
    plt.plot(episode_list, return_list, label=str(n_planning) + 'planning steps')

plt.legend()
plt.xlabel('Returns')
plt.title('Dyna-Q on {}'.format('Cliff Walking'))
plt.show()
```
