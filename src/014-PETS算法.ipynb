```python
import torch
import matplotlib.pyplot as plt
import r2l
import gymnasium as gym
import rl_utils
```

```python
buffer_size = 10000
n_sequence = 50
elite_ratio = 0.2
plan_horizon = 25
num_episodes = 10
env_name = 'Pendulum-v1'
env = gym.make(env_name)
device = r2l.get_gpu()
```

```python
pets = r2l.PETS(
    env,
    buffer_size,
    n_sequence,
    elite_ratio,
    plan_horizon,
    num_episodes,
    device
)
return_list = pets.train()
```

```python
episodes_list = list(range(len(return_list)))
plt.plot(episodes_list, return_list)
plt.xlabel('Episodes')
plt.ylabel('Returnx')
plt.title('PETS on {}'.format(env_name))
plt.show()
```
